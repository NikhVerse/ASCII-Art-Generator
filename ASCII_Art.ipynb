# ASCII ART: Kayadu-Lohar

```python
from PIL import Image
from google.colab import files

# Step 1: Upload the image
print("Please upload your image file:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Step 2: Settings (Change 'width' if you want it bigger/smaller)
width = 70
# Characters from dark (@) to light (.)
chars = ["@", "#", "S", "%", "?", "*", "+", ";", ":", ",", ".", " "]

# Step 3: Process the image
img = Image.open(filename).convert('L')  # Convert to black & white
w, h = img.size
aspect_ratio = h / w
new_height = int(aspect_ratio * width * 0.55) # 0.55 corrects for line height
img = img.resize((width, new_height))

# Step 4: Convert pixels to text
pixels = img.getdata()
# Map pixel brightness (0-255) to our list of characters
new_pixels = [chars[pixel // 25] for pixel in pixels]
new_pixels = "".join(new_pixels)

# Step 5: Print the result
ascii_image = "\n".join([new_pixels[index:(index + width)] for index in range(0, len(new_pixels), width)])
print("\n" + ascii_image)
```
